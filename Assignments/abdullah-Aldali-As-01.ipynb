### question 1


```python
cities_data = [
    {"city": "Damascus", "temperature": 38.5, "humidity": 30, "rainfall": 0.0},
    {"city": "Aleppo", "temperature": 41.0, "humidity": 25, "rainfall": 0.0},
    {"city": "Latakia", "temperature": 29.0, "humidity": 65, "rainfall": 12.5},
    {"city": "Homs", "temperature": None, "humidity": 40, "rainfall": 2.0},  # Missing temperature
    {"city": "Daraa", "temperature": 33.2, "humidity": 35, "rainfall": 1.5}
]
```

```python
valid_temperatures = []
valid_observations = 0

print("--- City Information & Initial Display ---")
for item in cities_data:
    # Check if temperature is missing
    if item["temperature"] is None:
        print(f"Skipping {item['city']}: Missing temperature data.")
        continue

    print(f"City: {item['city']} | Temp: {item['temperature']}°C | Humidity: {item['humidity']}% | Rainfall: {item['rainfall']}mm")
```

```python
classification_results = {}

for item in cities_data:
    if item["temperature"] is None:
        continue

    temp = item["temperature"]
    city_name = item["city"]

    # Classify temperatures according to the assignment thresholds
    if temp >= 40:
        category = "Extreme Heat"
    elif temp >= 35:
        category = "High Heat"
    elif temp >= 30:
        category = "Moderate Heat"
    else:
        category = "Normal"

    classification_results[city_name] = category
    valid_temperatures.append(temp)
    valid_observations += 1
```

```python
# Calculations using built-in functions[cite: 1]
avg_temp = round(sum(valid_temperatures) / len(valid_temperatures), 2)
highest_temp = max(valid_temperatures)
lowest_temp = min(valid_temperatures)

print("\n--- Climate Summary Statistics ---")
print(f"Number of valid observations: {valid_observations}")
print(f"Average temperature: {avg_temp}°C")
print(f"Highest temperature: {highest_temp}°C")
print(f"Lowest temperature: {lowest_temp}°C")

print("\n--- Final Classification Results Dictionary ---")
print(classification_results)
```

### question 2


```python
def calculate_risk(temperature, rainfall):
    if temperature >= 40:
        return "EXTREME"
    elif temperature >= 35 or rainfall < 5:
        return "HIGH"
    elif temperature >= 30 or rainfall < 15:
        return "MODERATE"
    else:
        return "LOW"
```

```python
# Assuming cities_data from Exercise 1 is already defined
print("--- Climate Risk Assessment for All Valid Cities ---")
for city in cities_data:
    if city["temperature"] is None:
        continue
    risk = calculate_risk(city["temperature"], city["rainfall"])
    print(f"City: {city['city']} | Temp: {city['temperature']}°C | Risk: {risk}")
```

```python
def get_climate_statistics(cities):
    # Extract only valid temperatures
    valid_temps = [c["temperature"] for c in cities if c["temperature"] is not None]

    avg_temp = round(sum(valid_temps) / len(valid_temps), 2)
    min_temp = min(valid_temps)
    max_temp = max(valid_temps)

    return (avg_temp, min_temp, max_temp)

# Call the function and print the tuple
stats_tuple = get_climate_statistics(cities_data)
print("\n--- Statistics Tuple (Average, Minimum, Maximum) ---")
print(stats_tuple)
```

```python
print("\n--- Cities with HIGH or EXTREME Risk ---")
for city in cities_data:
    if city["temperature"] is None:
        continue
    risk = calculate_risk(city["temperature"], city["rainfall"])
    if risk in ["HIGH", "EXTREME"]:
        print(f"City: {city['city']} | Temp: {city['temperature']}°C | Risk: {risk}")
```

```python
# Filter out the city with None temperature first to avoid sorting errors
valid_cities = [c for c in cities_data if c["temperature"] is not None]

# Sort using sorted() and lambda, setting reverse=True for highest to lowest
sorted_cities = sorted(valid_cities, key=lambda x: x["temperature"], reverse=True)

print("\n--- Cities Sorted by Temperature (Highest to Lowest) ---")
for city in sorted_cities:
    print(f"City: {city['city']} | Temp: {city['temperature']}°C")
```

### question 3


```python
# Parent Class
class ClimateStation:
    def __init__(self, city, temperature, humidity, rainfall):
        self.city = city
        self.temperature = temperature
        self.humidity = humidity
        self.rainfall = rainfall

    def display_summary(self):
        print(f"Station: {self.city} | Temp: {self.temperature}°C | Humidity: {self.humidity}% | Rainfall: {self.rainfall}mm")

    def calculate_risk(self):
        if self.temperature is None:
            return "UNKNOWN"
        if self.temperature >= 40:
            return "EXTREME"
        elif self.temperature >= 35 or self.rainfall < 5:
            return "HIGH"
        elif self.temperature >= 30 or self.rainfall < 15:
            return "MODERATE"
        else:
            return "LOW"

    def update_temperature(self, new_temperature):
        self.temperature = new_temperature
        print(f"-> Updated {self.city}'s temperature to {self.temperature}°C")

    # Bonus: Recommendation method
    def recommendation(self):
        risk = self.calculate_risk()
        if risk == "LOW":
            return "Normal monitoring"
        elif risk == "MODERATE":
            return "Continue monitoring"
        elif risk == "HIGH":
            return "Increased monitoring recommended"
        elif risk == "EXTREME":
            return "Immediate attention required"
        else:
            return "No data available"

# Child Class
class SmartClimateStation(ClimateStation):
    def __init__(self, city, temperature, humidity, rainfall, sensor_status="Active"):
        super().__init__(city, temperature, humidity, rainfall)
        self.sensor_status = sensor_status

    def check_sensor(self):
        print(f"Smart Sensor Status for {self.city}: [ {self.sensor_status} ]")


# --- Execution ---
print("=== 1. Creating and Displaying 5 Climate Stations ===")
stations = [
    ClimateStation("Damascus", 38.5, 30, 0.0),
    ClimateStation("Aleppo", 41.0, 25, 0.0),
    ClimateStation("Latakia", 29.0, 65, 12.5),
    ClimateStation("Homs", 34.0, 40, 2.0),
    ClimateStation("Daraa", 33.2, 35, 1.5)
]

for station in stations:
    station.display_summary()
    print(f"Risk: {station.calculate_risk()} | Recommendation: {station.recommendation()}\n")


print("=== 2. Testing Update Temperature Method ===")
stations[0].update_temperature(42.0)
print(f"New Risk for {stations[0].city}: {stations[0].calculate_risk()} | Action: {stations[0].recommendation()}\n")


print("=== 3. Testing Smart Climate Station (Child Class) ===")
smart_station = SmartClimateStation("Damascus-Smart", 41.5, 28, 0.0, sensor_status="Optimal")
smart_station.display_summary()  # Inherited method
smart_station.check_sensor()     # Child class method
print(f"Risk: {smart_station.calculate_risk()} | Recommendation: {smart_station.recommendation()}")
```
